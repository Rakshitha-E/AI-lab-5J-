```python

def is_variable(x):
    return isinstance(x, str) and x.startswith('?')


def occurs_check(var, expr):
    """Check whether variable occurs inside expression."""
    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs_check(var, item) for item in expr)

    return False


def substitute(expr, subst):
    """Apply substitutions recursively."""
    if is_variable(expr) and expr in subst:
        return substitute(subst[expr], subst)

    if isinstance(expr, tuple):
        return tuple(substitute(item, subst) for item in expr)

    return expr


def unify(x, y, subst=None):
    """Unification algorithm."""

    if subst is None:
        subst = {}

    x = substitute(x, subst)
    y = substitute(y, subst)

    # Step 1: If both are identical
    if x == y:
        return subst

    # Step 2: If x is a variable
    if is_variable(x):
        if occurs_check(x, y):
            return None
        subst[x] = y
        return subst

    # Step 3: If y is a variable
    if is_variable(y):
        if occurs_check(y, x):
            return None
        subst[y] = x
        return subst

    # Step 4: Both must be predicates
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        return None

    # Step 5: Check predicate names
    if x[0] != y[0]:
        return None

    # Step 6: Check number of arguments
    if len(x) != len(y):
        return None

    # Step 7: Unify corresponding arguments
    for a, b in zip(x[1:], y[1:]):
        subst = unify(a, b, subst)

        if subst is None:
            return None

    return subst


# Main program
print("UNIFICATION ALGORITHM")

p1 = ('P', '?x', 'a')
p2 = ('P', 'b', '?y')

result = unify(p1, p2)

if result is None:
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution Set:", result)
```
