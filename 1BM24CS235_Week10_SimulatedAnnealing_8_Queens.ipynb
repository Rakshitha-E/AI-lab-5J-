```python
import random
import math

def cost(board):
    conflicts = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                conflicts += 1

    return conflicts


def simulated_annealing(n):
    board = [random.randint(0, n - 1) for _ in range(n)]
    temperature = 100

    while temperature > 0.01:
        current_cost = cost(board)

        if current_cost == 0:
            return board

        next_board = board.copy()
        col = random.randint(0, n - 1)
        next_board[col] = random.randint(0, n - 1)

        next_cost = cost(next_board)
        delta = current_cost - next_cost

        if delta > 0:
            board = next_board
        else:
            probability = math.exp(delta / temperature)

            if random.random() < probability:
                board = next_board

        temperature *= 0.99

    return board


n = int(input("Enter number of queens: "))

while True:
    solution = simulated_annealing(n)

    if cost(solution) == 0:
        break

print("Solution:", solution)
print("Cost:", cost(solution))

for row in range(n):
    for col in range(n):
        if solution[col] == row:
            print("Q", end=" ")
        else:
            print(".", end=" ")
    print()

```
